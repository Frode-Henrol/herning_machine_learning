Vi bruger et datasæt fra Scikit-learn: wine
Du skal bruge Principal Component Analysis (PCA) til at reducere datasættets dimensioner fra 13 til et mere håndterbart niveau (f.eks. 2 eller 3 komponenter), samtidig med at du bevarer så meget af den originale varians som muligt.
Datasættet kan hentes med : sklearn.datasets.load_wine()
Reducer dimensionerne til 2 og undersøg hvor meget varians, der er forklaret af de 2 dimensioner
Gentag for 3 dimensioner

```python
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
wine = load_wine()

X = wine.data       # features
y = wine.target     # wine classes

X = StandardScaler().fit_transform(X)

pca1 = PCA(n_components=2)
X_pca1 = pca1.fit_transform(X)

print(pca1.explained_variance_ratio_)

pca2 = PCA(n_components=3)
X_pca2 = pca2.fit_transform(X)

print(pca2.explained_variance_ratio_)
```

```python
def model_training(model,data):
    model.fit(data)
    pred = model.transform(data)
    #plt.scatter(pred[:,0],pred[:,1])
    return pred
```

```python
pred1 = model_training(pca1, wine.data)
pred1.shape
```

```python
pca1.explained_variance_ratio_
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3)
kpred = model_training(model, pred1)
res = model.predict(pred1)
res
```
