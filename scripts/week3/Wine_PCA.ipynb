```python
from sklearn import datasets
```

```python
import matplotlib.pyplot as plt
```

```python
wine = datasets.load_wine()
```

```python
print(wine.DESCR)
```

```python
wine.feature_names
```

```python
from sklearn.decomposition import PCA
```

```python
wine_model = PCA(n_components=2)
```

```python
def model_training(model,data):
    model.fit(data)
    pred = model.transform(data)
    #plt.scatter(pred[:,0],pred[:,1])
    return pred
```

```python
pred = model_training(wine_model,wine.data)
pred.shape
```

```python
wine_model.explained_variance_ratio_
```

```python
wine_model2 = PCA(n_components=3)
```

```python
pred2 = model_training(wine_model2,wine.data)
pred2.shape
```

```python
wine_model2.explained_variance_ratio_
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3) 
```

```python
kpred = model_training(model,pred)
```

```python
res = model.predict(pred)
res
```
